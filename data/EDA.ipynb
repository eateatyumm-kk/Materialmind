```python
pd.set_option("display.max_columns", None)

data2 = pd.read_csv("materials_data.csv")
data2.head()
```

```python

data2_cols =[]

for col in data2.columns:
    if "mean" in col:
        data2_cols.append(col)
    
data2_mean = data2[data2_cols]

data2_mean.columns = [col.replace("MagpieData ", "") for col in data2_mean.columns]

corr_matrix = data2_mean.corr(numeric_only=True)

plt.figure(figsize=(14, 10)) 
sns.heatmap(
    corr_matrix, 
    annot=True,       
    fmt=".2f",        
    cmap='coolwarm',  
    vmin=-1, vmax=1,  
    cbar=True
)

plt.title("Correlation Matrix of Mean Magpie Properties", fontsize=14)
plt.xticks(rotation=45, ha='right')  
plt.tight_layout()
plt.show()
```
